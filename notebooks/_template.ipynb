# Notebook template

This is the starting point for every analysis notebook in the `notebooks/` lane.
Copy it, rename it to `NN-topic.ipynb`, and replace the cells below. The
conventions and mechanics are in [`README.md`](README.md).

**Rule:** logic that a second notebook would need does not live in a cell. It goes
in an importable module beside the notebooks, so it can be tested and cannot
diverge between copies. Cells are for the analysis you are showing, not for the
library you are building.


## Load the committed sample

The fixture path is relative to **this notebook**, because `nbmake` executes a
notebook with the notebook's own directory as the working directory. From
`notebooks/`, the committed sample is `../data/samples/`. An absolute path
(`C:\...`, `/Users/...`) breaks for every other clone and for the judges, and a
repository-root path silently resolves to the wrong place. `data/samples/` holds
synthetic data only; never read `data/raw/` from a committed notebook.


```
import pandas as pd

# Notebook-relative: nbmake runs this notebook with notebooks/ as its working
# directory, so the path is written relative to the notebook, not to the
# repository root.
indicators = pd.read_csv("../data/samples/public_health_indicators.csv")

indicators.head()
```

## Summarise

A simple group-by over `indicator` and `value`, showing the shape of the analysis
a real notebook would carry. Extend this cell rather than starting a second one.


```
by_indicator = indicators.groupby("indicator")["value"].mean()

for indicator, mean_value in by_indicator.items():
    print(f"{indicator}: {mean_value:.1f}")
```
